```python
adult_df['workclass'].value_counts()
workclass_df=adult_df['workclass'].value_counts().keys()
count=adult_df['workclass'].value_counts().values
work_data=pd.DataFrame(zip(workclass_df,count),
                      columns=['workclass','count'])
work_data
plt.figure(figsize=(14,6))
plt.bar('workclass','count',data=work_data)
plt.title("Barchart")
plt.xlabel("workclass")
plt.ylabel("count")
plt.savefig("D:\\course videos\\naresh it ds\\python-sessions-2\\workclass_info.jpg")
plt.show()
```

```python
adult_df['workclass'].value_counts()
workclass_df = adult_df['workclass'].value_counts().index
count = adult_df['workclass'].value_counts().values
work_data = pd.DataFrame(zip(workclass_df, count), columns=['workclass', 'count'])
print(work_data)
plt.figure(figsize=(14,6))
plt.bar(work_data['workclass'], work_data['count'])
plt.title("Barchart")
plt.xlabel("workclass")
plt.ylabel("count")
plt.xticks(rotation=45)
plt.yticks(rotation=45)# Rotate the x labels for better readability
plt.show()
```
