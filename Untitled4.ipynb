```python
import pywhatkit as kit
import time
import styletext
import random

# Replace with the recipient's WhatsApp number, including country code
phone_number = "91 9491201336"

# Number of times to send the message
count = 5

for i in range(1, count + 1):
    # Generate a random font style
    font = random.choice(list(styletext.FontStyle))
    styled_message = styletext.Text(message="By using Python code, I am preparing a MSG", font=font)
    message = f"{styled_message} for {i} {'time' if i == 1 else 'times'} repetition"
    
    # Send the message instantly
    kit.sendwhatmsg_instantly(phone_number, str(styled_message))
    # Adjust the time interval if needed
    time.sleep(5)
```
