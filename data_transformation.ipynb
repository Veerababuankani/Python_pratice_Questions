```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=np.random.exponential(size=10000)
```

```python
log_data=np.log(data)
plt.hist(log_data,bins=40,label='Log transformation')
plt.legend()
plt.show()
```

```python
 plt.subplot(1,2,1).hist(data,bins=40)
plt.subplot(1,2,2).hist(log_data,bins=40)
plt.legend()
plt.show()
```

```python
log_data=np.log(log_data)
plt.hist(log_data,bins=40,label='Log transformation')
plt.legend()
plt.show()
```

```python
log_data=1/data
plt.hist(log_data,bins=40,label='reciprocal transformation')
plt.legend()
plt.show()
```

```python
squt_data=np.sqrt(data)
plt.hist(squt_data,bins=40,label='square root transformation')
plt.legend()
plt.show()
```
