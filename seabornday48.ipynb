```python
#importing packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

####################### reading the data #########################################
#read the data from the file of csv
adult_df=pd.read_csv("D:\\ADITI DIGITAL SOLUTIONS-DS\\ds notes\\adult.csv")
#adult_df

################ Creating the frequncy table ################################

#method-1
adult_df["workclass"].value_counts()
workclass=adult_df["workclass"].value_counts().keys()
count=adult_df["workclass"].value_counts().values
workclass_data=pd.DataFrame(zip(workclass,count),columns=["workclass","count"])
workclass_data

################# Drawing thematplotlib.pyplot plot ################################

plt.figure(figsize=(14,6))
plt.bar('workclass','count',data=workclass_data)
plt.title("Barplot")
plt.xlabel("workclass")
plt.ylabel("count")
plt.show()
```
