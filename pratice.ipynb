```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path="D:\\course videos\\naresh it ds\\python-sessions-2\\adult.csv"
adult_df=pd.read_csv(file_path)
adult_df
```

```python
column_names=adult_df.columns
column_names
```

```python
#using matplotlib and pandas
keys=adult_df["workclass"].value_counts().keys()
count=adult_df["workclass"].value_counts().values
data_df=pd.DataFrame(zip(keys,count),columns=['workclass','count'])
data_df
###########plt######
plt.figure(figsize=(14,6))
plt.bar("workclass","count",data=data_df)
plt.xlabel("workclass")
plt.ylabel("count")
plt.show()
```

```python
#seaborn
sns.countplot(adult_df,x="workclass")
```

```python
labels=adult_df['workclass'].value_counts().keys()
sns.countplot(adult_df,
              x="workclass",
             order=labels)
plt.title("Barchart")
plt.show()
```

```python
count=adult_df['workclass'].value_counts()
ax=count.plot(kind='bar')
ax.bar_label(ax.containers[0])
plt.title("BarChart")
plt.show()
```

```python
#column_names
for i in column_names:
    #using matplotlib and pandas
    value1=adult_df[i].value_counts().keys()
    value2=adult_df[i].value_counts().values
    data_df=pd.DataFrame(zip(value1,value2),columns=[i,'count'])
    data_df
    ###########plt######
    plt.figure(figsize=(14,6))
    plt.bar(i,"count",data=data_df)
    plt.title(i)
    plt.xlabel(i)
    plt.ylabel("count")
    plt.show()
    ############################################
    plt.pie(x=value2,labels=value2,autopct="%0.2f%%",shadow=True,startangle=180,radius=2)
    plt.show()
```
