step-1 : finding the outliers

```python
for i in num_data:
    #calucating the q1,q2,q3 values
    q1=np.quantile(train_df[i],0.25)
    q2=np.quantile(train_df[i],0.50)
    q3=np.quantile(train_df[i],0.75)
    #Calucating the IQR(q3-q1)
    IQR=q3-q1
    #calucating the Uperbound(ub)
    ub=q3+1.5*IQR
    #calucating the lowerbound(lb)
    lb=q1-1.5*IQR
    #con1=col>ub
    #con2=col2<lb
    col1=train_df[i]>ub
    col2=train_df[i]<lb
    #outliers
    outliers=train_df[i][col1|col2]
    #series into arry of values by applying a value
    outliers_data=outliers.values
    print(len(outliers_data))
    
```

```python
for i in num_data:
    # Calculating the q1, q2, q3 values
    q1 = np.percentile(train_df[i], 25)
    q2 = np.percentile(train_df[i], 50)
    q3 = np.percentile(train_df[i], 75)
    
    # Calculating the IQR (q3 - q1)
    IQR = q3 - q1
    
    # Calculating the Upperbound (ub)
    ub = q3 + 1.5 * IQR
    
    # Calculating the Lowerbound (lb)
    lb = q1 - 1.5 * IQR
    
    # Filtering non-outliers
    col1 = train_df[i] <= ub  # Fixed logical comparison
    col2 = train_df[i] >= lb  # Fixed logical comparison
    non_outliers = train_df[i][col1 | col2]  # Use '&' for logical AND

    # Series into an array of values by applying a value
    non_outliers_data = non_outliers.values  # Corrected variable usage
    print(len(non_outliers_data))  # Updated variable to match

    # Plotting the 'non_outliers'
    plt.figure(figsize=(10, 8))
    
    # Subplot 1: Histogram of non-outliers
    plt.subplot(2, 2, 1)
    plt.title("Non-Outliers"+"  "+i)
    plt.hist(non_outliers, bins=9)  # Corrected indexing

    
    # Subplot 2: Histogram of original data
    plt.subplot(2, 2, 2)
    plt.title('Original Data'+"  "+i)
    plt.hist(train_df[i], bins=9)
    
    # Subplot 3: Boxplot of non-outliers
    plt.subplot(2, 2, 3)
    plt.title('Non-Outliers'+"   "+i)
    plt.boxplot(non_outliers, vert=False)
    
    # Subplot 4: Boxplot of original data
    plt.subplot(2, 2, 4)
    plt.title('Original'+"   "+i)
    plt.boxplot(train_df[i], vert=False)
```
