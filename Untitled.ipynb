```python
#importing packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
adult_df=pd.read_csv("D:\\ADITI DIGITAL SOLUTIONS-DS\\ds notes\\adult.csv")
en_count=round(adult_df['education-num'].count(),2)
en_min=round(adult_df['education-num'].min(),2)
en_max=round(adult_df['education-num'].max(),2)
en_mean=round(adult_df['education-num'].mean(),2)
en_median=round(adult_df['education-num'].median(),2)
en_std=round(adult_df['education-num'].std(),2)
val_minus1=round(en_mean-1*en_std,2)
val_plus1=round(en_mean+1*en_std,2)
#95
val_minus2=round(en_mean+2*en_std,2)
val_plus2=round(en_mean+2*en_std,2)
#99
val_minus3=round(en_mean-3*en_std,2)
val_plus3=round(en_mean+3*en_std,2)
print(val_minus1,val_plus1,val_minus2,val_plus2,val_minus3,val_plus3)

con1=adult_df['education-num']>val_minus1
con2=adult_df['education-num']>val_plus1
print(len(adult_df[con1&con2]))
print(len(adult_df[con1&con2])/len(adult_df))
```
