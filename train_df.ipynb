Drawing the frequency and making the barchart,piechart

```python
for i in cat_data:
    col=train_df[i].value_counts().keys()
    count=train_df[i].value_counts().values
    fre_data=pd.DataFrame(zip(col,count),columns=[i,'count'])
    print(fre_data)
    plt.figure(figsize=(14,6))
    #bar graph
    plt.subplot(1,2,1)
    plt.bar(fre_data[i], fre_data['count'])
    plt.title('Bar Graph of {}'.format(i))
    plt.xlabel(i)
    plt.ylabel("count")
    plt.savefig("D:\\course videos\\naresh it ds\\python-sessions-2\\train_ctrUa4K sub file with plot\\{} bargraph.jpg".format(i))
    #pie chart
    plt.subplot(1,2,2)
    plt.title('pie Chart of {}'.format(i))
    plt.pie(x=count,
            labels=col,
            autopct='%0.2f%%',
            shadow=True,
            startangle=180,
            radius=1)
    plt.savefig("D:\\course videos\\naresh it ds\\python-sessions-2\\train_ctrUa4K sub file with plot\\{} piechart.jpg".format(i))
    plt.show()
```
